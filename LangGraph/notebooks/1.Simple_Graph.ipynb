```python
from typing_extensions import TypedDict

class State(TypedDict):
    graph_info :str
```

```python
def start_play(state: State) -> None:
    """
    Start the game by displaying the graph information.
    """
    print("Starting the game...")
    return {"graph_info": state["graph_info"] + "I am planning to play a game."}
    # Add more game logic here
    
def cricket(state: State) -> None:
    """
    Play a cricket game.
    """
    print("Playing cricket...")
    return {"graph_info": state["graph_info"] + "I am playing cricket."}
    # Add more cricket logic here
    
def football(state: State) -> None:
    """
    Play a football game.
    """
    print("Playing football...")
    return {"graph_info": state["graph_info"] + "I am playing football."}
    # Add more football logic here
```

```python
import random 
from typing import Literal

def random_game(state: State) -> Literal["cricket", "football"]:
    """
    Randomly choose a game to play.
    """
    if random.random() < 0.5:
        return "cricket"
    else:
        return "football"
```

```python
from IPython.display import display, Image
from langgraph.graph import StateGraph, START, END

graph = StateGraph(State)

graph.add_node("start_play", start_play)
graph.add_node("cricket", cricket)
graph.add_node("football", football)

graph.add_edge(START, "start_play")
graph.add_conditional_edges("start_play", random_game)
graph.add_edge("cricket", END)
graph.add_edge("football", END)

graph_builder = graph.compile()

# display(Image(graph_builder.get_graph().draw_mermaid_png(), width=400, height=300))
```
